```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import norm
from scipy.stats import ttest_ind
from scipy.stats import chi2_contingency
from scipy.stats import f_oneway
```

```python
# Load dataset
data = pd.read_excel("C:/Users/MAHESH LOHAR/ai python in class/Dataset_File/health_dataset_1000.xlsx")

data.head()
```

```python
#Dataset Information
data.info()
print("---------------------------------------------------")
# kitne null value hai
data.isnull().sum()
```

```python
# Task 2 - Confidence Interval for Age

mean = data["age"].mean()
std = data["age"].std()

n = len(data)
z = 1.96

lower = mean - z*(std/np.sqrt(n))
upper = mean + z*(std/np.sqrt(n))

print("95% Confidence Interval")
print(lower)
print(upper)
```

```python
#Task 3 - Critical Value and P-value

critical_value = norm.ppf(0.975)
print("Critical Value =",critical_value)
```

```python
# Task 4 - T-Test #
#Compare BMI of Smokers and Non-Smokers.

smoker = data[data["smoking_status"]=="Smoker"]["bmi"]
non = data[data["smoking_status"]=="Non-Smoker"]["bmi"]
t,p = ttest_ind(smoker,non)

print(t)
print(p)
```

```python
# Task 5 - Chi-Square Test

# Relationship between Smoking Status and Diabetes.

table = pd.crosstab(data["smoking_status"],data["diabetes"])
chi2,p,dof,expected = chi2_contingency(table)

print(chi2)
print(p)
```

```python
# Task 6 - ANOVA

# Compare BMI among different exercise groups.

daily = data[data["exercise_frequency"]=="Daily"]["bmi"]
weekly = data[data["exercise_frequency"]=="Weekly"]["bmi"]
rarely = data[data["exercise_frequency"]=="Rarely"]["bmi"]
never = data[data["exercise_frequency"]=="Never"]["bmi"]

f,p = f_oneway(daily,weekly,rarely,never)

print(f)
print(p)
```

```python
#Task 7

#Covariance

cov = data["age"].cov(data["bmi"])
print("Covariance - ",cov)

#Correlation

corr = data["age"].corr(data["bmi"])
print("Correlation - ",corr)
```
