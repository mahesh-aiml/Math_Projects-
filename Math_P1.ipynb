5 . Contingency Table & Probability Calculations

```python
# Task 1
contingency_table = pd.crosstab(
    data["group_discussion"],
    data["final_exam_pass"])
print(contingency_table)
joint_count = len(data[
                  (data["group_discussion"] == "Yes") &
                  (data["final_exam_pass"] == "Pass")])
print("--------------------------------------------------------")

#Task : 2
# Q 1
joint_probability = joint_count / len(data)
print("Joint Count =", joint_count)
print("Joint Probability =", round(joint_probability,3))


print("--------------------------------------------------------")
# Q 2
pass_count = (data["final_exam_pass"] == "Pass").sum()
marginal_probability = pass_count / len(data)
print("Pass Count =", pass_count)
print("Marginal Probability =", round(marginal_probability,3))
```

6 : Understanding Relationships

```python
# Task : 1

overall = (data["final_exam_pass"] == "Pass").mean()

# Conditional Probability
conditional = len(data[
                  (data["group_discussion"] == "Yes") &
                  (data["final_exam_pass"] == "Pass")]) / len(data[data["group_discussion"] == "Yes"])
print("Overall Probability =", round(overall,3))
print("Conditional Probability =", round(conditional,3))
print("--------------------------------------------------------")

# Task : 2
if conditional > overall:
    print("Dependent Events")
elif abs(conditional - overall) < 0.01:
    print("Independent Events")
else:
    print("Dependent Events")

print("Not Mutually Exclusive")
```
